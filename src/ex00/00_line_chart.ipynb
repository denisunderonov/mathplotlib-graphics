# Линейный график

```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect('../data/checking-logs.sqlite')
```

```python
query = """
select * from pageviews where uid like 'user_%'
"""

df = pd.read_sql(query, conn)
```

```python
query = """
SELECT strftime('%Y-%m-%d', datetime) AS day, COUNT(*) as views
FROM pageviews
WHERE uid LIKE 'user_%'
GROUP BY day
ORDER BY day;
"""

cnt_visits = pd.read_sql(query, conn)
cnt_visits
```

```python
cnt_visits.plot(
    x='day',
    y='views',
    title="Views per day",
    figsize=(15,8),
    rot=90,
    fontsize=8
)
```

```python
conn.close()
```
