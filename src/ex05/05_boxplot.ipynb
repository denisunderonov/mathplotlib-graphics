# Диаграмма размаха

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/ab-test.csv')
df
```

```python
fig, axes = plt.subplots(
    1, 2,
    figsize=(15, 8),
    sharey=True
)

periods = ['after', 'before']

for ax, period in zip(axes, periods):
    period_data = df[df['time'] == period]

    period_data.boxplot(
        column='diff',  
        by='group',      
        ax=ax,

        boxprops=dict(color='green', linewidth=3),
        medianprops=dict(color='red', linewidth=2),
        whiskerprops=dict(color='indianred', linewidth=1.5),
        capprops=dict(color='indianred', linewidth=1.5),
        showmeans=True,
        meanprops=dict(marker='^', markerfacecolor='forestgreen',
                       markeredgecolor='forestgreen', markersize=7)
    )

    ax.set_title(period)
    ax.set_xlabel('[group]')
    ax.set_facecolor('#EAEAF2')
    ax.grid(True, color='#B0B0B0', alpha=0.7)

fig.suptitle('Lag before labs deadlines. After and before the Newsfeed',
             fontsize=15)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

«Каков IQR (межквартильный размах) контрольной группы до появления ленты новостей?»

Ответ: приблизительно 150.
